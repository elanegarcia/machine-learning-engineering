```python
#carrega e inspeciona

import pandas as pd, ast, numpy as np

df = pd.read_csv("../data/dataset_rec.csv")
df.shape                   # (100, 3)
df.isna().sum()            # checa nulos
df.url.duplicated().sum()  # checa URLs duplicadas
```

```python
# Parsear os embeddings e checa consistência

E = np.array([ast.literal_eval(e) for e in df.embedding])
E.shape  # (100, 768) — confirma que todas as linhas tem a mesma dimensão
```

```python
# Checa a geometria do espaço vetorial

norms = np.linalg.norm(E, axis=1)
norms.min(), norms.mean(), norms.max()   # os vetores estão normalizados? (aqui, não)

X = E / norms[:, None]
S = X @ X.T
np.fill_diagonal(S, np.nan)
np.nanmean(S), np.nanmin(S), np.nanmax(S)  # cosseno médio muito alto (~0.66) indica anisotropia
```
